## Disease Background: 

1. Prevalence & incidence

    The prevalence of FSHD is about 21,000 individuals in the United States. In the world, the prevalence rate is estimated to be between 2.03 and 6.8 for every 100,000 individuals. The global incidence rate may be anywhere from 1 in 7500 births to 1 in 14,286 births. All together, this is a rare disease.

2. Symptoms

    FSHD is characterized by muscle atropy, leading to weakness and an inability to do common everyday tasks (e.g. climbing up stairs). "Foot drop" is a textbook symptom, as FSHD tends to favorably affect the tibialis anterior muscle which holds our feet up. The biological process of muscle atropy (toxic DUX4 expression) may also induce inflammation; patients with FSHD commonly report chronic pain in their shoulders, neck, lower back, and legs. Interstingly, many patients with FSHD also have an abornmal twisting or dilation of blood vessels in their retinas. Other, more mild and rare problems include hearing loss, cardiac arrythmias, and difficulty breathing. All these symptoms may cascade into misalignment of spine muscle loss/pain and unfortunately may cause embarassment and emotional distress for the patient.

    ---

3. Diagnosis
    
    Typically, patients with a bizarre muscular weakness may be recommended for a genetic or blood test for FSHD. FSHD1 is the most common (95%) and dominantly inherited. It is characterized by a chunk of repeats absent at the 4q35 region at the tip of chromosome 4; people with FSHD has fewer than 10 repeats of that segment. Least common is FSHD2 (5%), which is recessively inherited, so both parents must have carry the gene for it to (maybe) pass on to their offspring. Patients with FSHD2 have a normal level of repeats at 4q35 on chromosome 4, but a mutation of gene SMCHD1 on chromosome 18 means that their 4q35 is less methylated and thus more strung out and loosened. 

        https://www.urmc.rochester.edu/neurology/fshd-center/patients 
        https://www.fshdsociety.org/living-with-fshd/understanding-fshd/genetic-cause/

4. Standard of care treatments (& reimbursement) 
    
    Unfortuantely, there really is no (not yet!) "magic pill" for treating the root cause of FSHD, but there are some remedies to treat its underlying symptoms. For instance, foot braces may help with patients who tend to trip or have trouble walking due to the weakening of their tibialis anterior muscle. Physical therapy helps with keeping joint mobility, alongside surgical scapular fixation for arm mobility. Other treatments include hearing aids for those with hearing loss, BIPAP machines for those who have difficulty breathing while laying down to sleep, and pain meds. Insurance covers an average of $19,370 per FSHD patient annually and Medicare reimburses $11,704.

        https://pubmed.ncbi.nlm.nih.gov/42325276/ 
        https://www.urmc.rochester.edu/neurology/fshd-center/patients

5. Disease progression & prognosis

    FSHD generally progresses slowly, although its severity and progression vary significantly from patient to patient. Symptoms usually begin during adolescence, impacting muscles in the face, shoulders, and arms. Weakness may later progress to affect the hips and legs. Muscle weakening is oftentimes asymmetric, impacting one side more than the other, fostering imbalance and furthar muscle weakness. Approximately 20% of patients will require a wheelchair as the disease progresses. While the life expectancy is generally normal, FSHD can severely weaken respiratory muscle and occasionally cardiac muscle, causing serious complications.

        https://pubmed.ncbi.nlm.nih.gov/20301616/
        https://www.fshdsociety.org/living-with-fshd/managing-fshd/clinical-care/

6. Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)

    FSHD is caused by an error when expressing DUX4, a gene active during development but silenced in mature skeletal muscle. In FSHD, reduced gene silencing allows DUX4 to activate an embryonic gene program in mature muscle cells, interupting their normal function and damaging muscle. At the cellular and molecular level, damaged muscle is replaced by fat and connective tissue, reducing the muscle’s ability to generate force due to less over muscle fibers. Physiological examples of FSHD include: shoulder blades muscle weakness causing scapular winging and difficulty raising the arms and weakness of the tibialis anterior muscle in the leg causing foot drop.

        https://academic.oup.com/hmg/article/33/8/698/7588694
        https://pubmed.ncbi.nlm.nih.gov/42325276/

In June 2026, Novartis reported that the trial's Phase 1/2 biomarker met endpoints measuring a DUX4-regulated biomarker and creatine kinase, a marker of muscle damage. A Phase 3 trial, FORTITUDE-3, is evaluating safety and clinical efficacy. These results are steps in the right direction, but are yet to establish long-term positive effect.

## MRI Data Analysis Methdology:

Water and fat MRI images from subject UW_0020 were analyzed using the supplied class workflow. Eight cross-sectional locations were available at each of two visits (TP1 and TP2). TP1 was used for the primary analysis, with the class notebook’s optional longitudinal analysis included separately. Python, NumPy, pandas, Matplotlib, and pathlib were used. # 1. Loading the files

The metadata table, slice_info.csv, linked each slice to its water, fat, and segmentation-mask files and recorded its anatomical location. Images were stored as NumPy arrays.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("FSHD_TA_MRI_data_0020_TP1_TP2")
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")
manifest
```

# 2. Identifying the tibialis anterior

The fourth TP1 slice (40% along the muscle, 92 mm from its start) was selected as the representative image. The supplied mask was converted to Boolean values: True identified TA muscle pixels and False identified other pixels. No new segmentation was performed. Pixel correspondence across water, fat, and mask arrays was assumed.

```python
row = manifest[manifest["timepoint"] == "TP1"].iloc[3]  # representative mid-muscle slice

water = np.load(DATA_DIR / row["water_file"])
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())
```

# 3. Calculating fat fraction

At each pixel, fat fraction (%) = 100 × fat signal / (water signal + fat signal). Division was performed where the combined signal was positive. Whole-region summaries included only those valid pixels. For the TA analysis, values outside the mask were replaced with NaN. The class implementation initializes nonpositive-denominator pixels to zero; an additional check confirmed that no TA pixels had nonpositive denominators in any supplied slice.

```python
denominator = water + fat
valid_pixels = denominator > 0

fat_fraction = np.divide(
    fat,
    denominator,
    out=np.zeros_like(fat, dtype=float),
    where=valid_pixels
) * 100

# Keep the whole supplied image region for the first analysis.
fat_fraction_full = fat_fraction.copy()

# Then apply the TA mask for the muscle-specific analysis.
fat_fraction_masked = fat_fraction.copy()
fat_fraction_masked[~mask] = np.nan

whole_image_values = fat_fraction_full[valid_pixels]
muscle_values = fat_fraction_masked[~np.isnan(fat_fraction_masked)]

print(f"Whole supplied image region — mean: {np.mean(whole_image_values):.1f}%")
print(f"Whole supplied image region — minimum: {np.min(whole_image_values):.1f}%")
print(f"Whole supplied image region — maximum: {np.max(whole_image_values):.1f}%")
print()
print(f"TA muscle only — mean: {np.mean(muscle_values):.1f}%")
print(f"TA muscle only — minimum: {np.min(muscle_values):.1f}%")
print(f"TA muscle only — maximum: {np.max(muscle_values):.1f}%")
```

# 4. Visualizing the muscle and fat-fraction map

Water and fat images were displayed with the TA boundary. Whole-region and cropped TA fat-fraction maps used the same 0–100% scale. Cropping changed the display only and did not change the region used for statistics. The cropped panel uses pixel coordinates relative to the crop.

```python
# Find the bounding box of the TA mask
y_pixels, x_pixels = np.where(mask)

y0, y1 = y_pixels.min(), y_pixels.max() + 1
x0, x1 = x_pixels.min(), x_pixels.max() + 1

# Crop the TA-specific fat-fraction map
fat_fraction_TA = fat_fraction_masked[y0:y1, x0:x1]

fig, axes = plt.subplots(1, 4, figsize=(15, 4))

# Water image with TA mask outline
axes[0].imshow(water, cmap="gray")
axes[0].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[0].set_title("Water image + TA mask")

# Fat image with TA mask outline
axes[1].imshow(fat, cmap="gray")
axes[1].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[1].set_title("Fat image + TA mask")

# Fat fraction for the whole supplied image region
whole_image = axes[2].imshow(
    fat_fraction_full,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[2].contour(mask, levels=[0.5], colors="cyan", linewidths=2)
axes[2].set_title("Fat fraction: whole region")

# Zoomed-in fat fraction for the TA only
ta_image = axes[3].imshow(
    fat_fraction_TA,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[3].set_title("TA fat fraction")

for ax in axes:
    ax.set_xlabel("Horizontal pixel position")
    ax.set_ylabel("Vertical pixel position")

# Both fat-fraction plots use the same 0–100% scale
fig.colorbar(
    ta_image,
    ax=axes[2:],
    fraction=0.046,
    pad=0.04,
    label="Fat fraction (%)"
)

plt.show()
```

# 5. Summarizing the distribution

Overlaid histograms with 30 bins for each region summarized the distribution of whole-region and TA-only pixel values. Raw pixel counts were used, so histogram heights depend on the number of pixels in each region. The spread of values provides information not captured by the mean alone.

```python
plt.figure(figsize=(7, 4))
plt.hist(whole_image_values, bins=30, alpha=0.55, label="Whole supplied region")
plt.hist(muscle_values, bins=30, alpha=0.65, label="TA muscle only")
plt.xlabel("Fat fraction (%)")
plt.ylabel("Number of pixels")
plt.title("Fat-fraction distributions")
plt.legend()
plt.show()
```

# 6. Summarizing sampled locations

The same function was applied to all eight TP1 slices. At each location, the mean, median, and population standard deviation (NumPy default ddof=0) of TA fat fraction were calculated, excluding NaN values. Muscle-pixel count was the sum of the Boolean mask. Standard deviation is expressed in percentage points. Each row retained subject, visit, slice ID, percentage location, and physical distance. These are cross-sectional summaries; no volume-weighted whole-muscle average was calculated.

```python
def calculate_fat_fraction(water, fat, mask):
    """Return a masked fat-fraction image in percent."""
    denominator = water + fat
    result = np.divide(
        fat,
        denominator,
        out=np.zeros_like(fat, dtype=float),
        where=denominator > 0
    ) * 100
    result[~mask] = np.nan
    return result

analysis_timepoint = "TP1"  # analyze one visit at a time
analysis_manifest = manifest[manifest["timepoint"] == analysis_timepoint]
results = []

for _, row in analysis_manifest.iterrows():
    water = np.load(DATA_DIR / row["water_file"])
    fat = np.load(DATA_DIR / row["fat_file"])
    mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
    ff = calculate_fat_fraction(water, fat, mask)

    results.append({
        "subject": row["subject"],
        "timepoint": row["timepoint"],
        "slice_id": row["slice_id"],
        "percent_along_muscle": row["percent_along_muscle"],
        "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
        "mean_fat_fraction_percent": np.nanmean(ff),
        "median_fat_fraction_percent": np.nanmedian(ff),
        "std_fat_fraction_percent": np.nanstd(ff),
        "n_muscle_pixels": int(mask.sum()),
    })

results_df = pd.DataFrame(results)
results_df
```

# 7. Optional comparison across visits

The class function was repeated for TP1 and TP2. Mean TA fat fraction was plotted against distance_from_muscle_start_mm because physical sampling intervals were not uniform and differed between visits. Connecting lines visualize the sampled profiles; they are not additional measurements or a separately calculated interpolation dataset.

```python
def summarize_visit(timepoint):
    visit_results = []
    visit_manifest = manifest[manifest["timepoint"] == timepoint]

    for _, row in visit_manifest.iterrows():
        water = np.load(DATA_DIR / row["water_file"])
        fat = np.load(DATA_DIR / row["fat_file"])
        mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
        ff = calculate_fat_fraction(water, fat, mask)

        visit_results.append({
            "timepoint": timepoint,
            "slice_id": row["slice_id"],
            "percent_along_muscle": row["percent_along_muscle"],
            "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
            "mean_fat_fraction_percent": np.nanmean(ff),
        })

    return pd.DataFrame(visit_results)

profiles_df = pd.concat(
    [summarize_visit("TP1"), summarize_visit("TP2")],
    ignore_index=True
)

plt.figure(figsize=(7, 4))
for timepoint, group in profiles_df.groupby("timepoint"):
    group = group.sort_values("distance_from_muscle_start_mm")
    plt.plot(
        group["distance_from_muscle_start_mm"],
        group["mean_fat_fraction_percent"],
        marker="o",
        label=timepoint
    )
plt.xlabel("Distance from muscle start (mm)")
plt.ylabel("Mean TA fat fraction (%)")
plt.title("Mean fat fraction along the tibialis anterior")
plt.legend(title="Visit")
plt.grid(True, alpha=0.3)
plt.show()
```

# 9. Matching locations between visits

For the optional paired comparison, mean_fat_fraction_percent was matched using percent_along_muscle. Change was calculated as TP2 minus TP1 in percentage points. This assumes the same percentage location represents approximately corresponding anatomical regions across visits, despite differences in physical distance. No image registration or interpolation was performed. Changes describe MRI composition and do not independently establish changes in patient function.


```python
comparison_source = profiles_df if "profiles_df" in globals() else results_df
comparison = comparison_source.pivot(
    index="percent_along_muscle",
    columns="timepoint",
    values="mean_fat_fraction_percent"
)

if {"TP1", "TP2"}.issubset(comparison.columns):
    comparison["change_TP2_minus_TP1"] = comparison["TP2"] - comparison["TP1"]
    print(comparison.to_string())
else:
    print("Run the optional two-visit extension before making a longitudinal comparison.")
```
