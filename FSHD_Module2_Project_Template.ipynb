## Project Goal:

Broadly, our goal is to determine: which biomarkers are best for testing a new drug's effectiveness for treating Facioscapulohumeral muscular dystrophy (FSHD)? Ideally, an exploratory endpoint consistently correlates well with disease progression and symptoms. As Professor Blemker emphasized during class, the decision to advance a drug into further clinical trials greatly depends on which variables are being measured. For instance, a measurement may correlate really well with the disease progression but does not detectably change enough over the two-year testing period to register any beneficial effect from the drug. Also, it's possible that this drug may show improvement in the patient's timed-up-and-go (TUG) test but does not actually slow the progression of the disease or alleviate any other symptoms. Moreover, a metric may overly fluctuate, making it harder to determine the genuine trend in disease progression. So, on that last note, we more specifically seek to understand: does MRI fat fraction or clinical severity score change more consistently over a two year period in patients with FSHD?

## Disease Background: 

1. Prevalence & incidence

    The prevalence of FSHD is about 21,000 individuals in the United States (Statland & Tawil, 2016). In the world, the prevalence rate is estimated to be between 2.03 and 6.8 for every 100,000 individuals. The global incidence rate may be anywhere from 1 in 7500 births to 1 in 14,286 births (UMass Chan Medical School, n.d.). All together, this is a rare disease. 

    ---

2. Symptoms

    FSHD is characterized by muscle atropy, leading to weakness and an inability to do common everyday tasks (e.g. climbing up stairs). "Foot drop" is a textbook symptom, as FSHD tends to favorably affect the tibialis anterior muscle which holds our feet up (Leuzinger-Dias et al., 2026). The biological process of muscle atropy (toxic DUX4 expression) may also induce inflammation; patients with FSHD commonly report chronic pain in their shoulders, neck, lower back, and legs (University of Rochester Medical Center, n.d.). Interstingly, many patients with FSHD also have an abornmal twisting or dilation of blood vessels in their retinas (Muscular Dystrophy Association, n.d.). Other, more mild and rare problems include hearing loss, cardiac arrythmias, and difficulty breathing (Fondazione Telethon, n.d.). All these symptoms may cascade into misalignment of spine muscle loss/pain and unfortunately may cause embarassment and emotional distress for the patient. 

    ---

3. Diagnosis
    
    Typically, patients with a bizarre muscular weakness may be recommended for a genetic or blood test for FSHD. FSHD1 is the most common (95%) and dominantly inherited (University of Rochester Medical Center, n.d.). It is characterized by a chunk of repeats absent at the 4q35 region at the tip of chromosome 4; people with FSHD has fewer than 10 repeats of that segment. Least common is FSHD2 (5%), which is recessively inherited, so both parents must have carry the gene for it to (maybe) pass on to their offspring. Patients with FSHD2 have a normal level of repeats at 4q35 on chromosome 4, but a mutation of gene SMCHD1 on chromosome 18 means that their 4q35 is less methylated and thus more strung out and loosened (FSHD Society, n.d.-b). 

    ---

4. Standard of care treatments (& reimbursement) 
    
    Unfortuantely, there really is no (not yet!) "magic pill" for treating the root cause of FSHD, but there are some remedies to treat its underlying symptoms. For instance, foot braces may help with patients who tend to trip or have trouble walking due to the weakening of their tibialis anterior muscle. Physical therapy helps with keeping joint mobility, alongside surgical scapular fixation for arm mobility. Other treatments include hearing aids for those with hearing loss, BIPAP machines for those who have difficulty breathing while laying down to sleep, and pain meds (Hill et al., 2026). Insurance covers an average of $19,370 per FSHD patient annually and Medicare reimburses $11,704 (University of Rochester Medical Center, n.d.).

    ---

5. Disease progression & prognosis

    FSHD generally progresses slowly, although its severity and progression vary significantly from patient to patient. Symptoms usually begin during adolescence, impacting muscles in the face, shoulders, and arms. Weakness may later progress to affect the hips and legs. Muscle weakening is oftentimes asymmetric, impacting one side more than the other, fostering imbalance and furthar muscle weakness (Preston & Wang, 2025). Approximately 20% of patients will require a wheelchair as the disease progresses (FSHD Society, n.d.-a). While the life expectancy is generally normal, FSHD can severely weaken respiratory muscle and occasionally cardiac muscle, causing serious complications.

    ___

6. Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)

    FSHD is caused by an error when expressing DUX4, a gene active during development but silenced in mature skeletal muscle. In FSHD, reduced gene silencing allows DUX4 to activate an embryonic gene program in mature muscle cells, interupting their normal function and damaging muscle. At the cellular and molecular level, damaged muscle is replaced by fat and connective tissue, reducing the muscle’s ability to generate force due to less over muscle fibers (Wong et al., 2024). Physiological examples of FSHD include: shoulder blades muscle weakness causing scapular winging and difficulty raising the arms and weakness of the tibialis anterior muscle in the leg causing foot drop (Hill et al., 2026).

    ___

7. Clinical Trials/next-gen therapies

    Delpacibart braxlosiran (developed by Novartis): A next-gen therapy which links an antibody to small interfering RNA (siRNA), delivering the siRNA into muscle cells to suppress DUX4 expression. In June 2026, Novartis reported that the trial's Phase 1/2 biomarker met endpoints measuring a DUX4-regulated biomarker and creatine kinase, a marker of muscle damage. A Phase 3 trial, FORTITUDE-3, is evaluating safety and clinical efficacy. These results are steps in the right direction, but are yet to establish long-term positive effect (Novartis, 2026).

    EPI-321 (developed by UCLA Health): A gene therapy which uses genetic editing (CRISPR) to silence DUX4 without changing the orginal DNA sequence. Its Phase 1/2 trial is evaluating safety and early measures of activity in adults with FSHD1. All 12 participants had been enrolled and dosed by July 2026, but a larger study could be effective (University of California Health, 2026).

## Data Set:

The FSHD dataset includes 34 subjects (16 female, 18 male), with ages ranging from 21 to 69 years old. Clinical, biomarker, and functional measurements are taken three times (baseline, year 1, and year 2) over roughly a 2 year period, with lag times recorded to adjust for follow-ups not taken on the exact annual date. They include dorsiflexion (measures strength of tibialis anterior muscle), as well as TUG (timed up and go test), timed water rescue (wr), ascend_4 (time to go up 4 stairs), and descend_4 (time to go down 4 stairs). Needle biopsies were performed to evaluate DUX4 expression. Each patient is also assigned a clinical severity score (10-point scale) (Wong et al., 2024). 

To analyze the MRI images, the researchers used an AI algorithm (from a company based in Charlottesville!) to add up the amount of fat infiltration and volume in each of 8 slices taken of the tibialis anterior, indexed by their distance from where the muscle starts (Wong et al., 2024). The MRI technique called "STIR" was used to detect inflammation by suppressing the high signal from fat (MRI Master, n.d.). Another  MRI technique called "Dixon" was used for detecting water and fat (Radiopaedia, n.d.). A segmentation mask labels the boundary that defines the location of the tibialis anterior muscle. One image shows the amount and location of water in the slice, with a lighter color to represent higher water concentration. The same set-up applies for the fat image, except here, the lighter color represents high fat concentration. The image showing fat fraction uses percentage units in giving the proportion of fat over the combined amount of fat and water (Wong et al., 2024). 

This data was collected through the Seattle Paul D. Wellstone Muscular Dystrophy Specialized Research Center and obtained at the University of Washington Medical Center, University of Rochester Medical Center, and Kansas University Medical Center. The original study is called "Regional and bilateral MRI and gene signatures in facioscapulohumeral dystrophy: implications for clinical trial design and mechanisms of disease progression" by Chao-Jen Wong and colleagues (Wong et al., 2024).

# 1. Loading the files

The metadata table, slice_info.csv, linked each slice to its water, fat, and segmentation-mask files and recorded its anatomical location. Images were stored as NumPy arrays.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("FSHD_TA_MRI_data_0020_TP1_TP2")
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")
manifest
```

# 2. Identifying the tibialis anterior

The fourth TP1 slice (40% along the muscle, 92 mm from its start) was selected as the representative image. The supplied mask was converted to Boolean values: True identified TA muscle pixels and False identified other pixels. No new segmentation was performed. Pixel correspondence across water, fat, and mask arrays was assumed.

```python
row = manifest[manifest["timepoint"] == "TP1"].iloc[3]  # representative mid-muscle slice

water = np.load(DATA_DIR / row["water_file"])
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())
```

# 3. Calculating fat fraction

At each pixel, fat fraction (%) = 100 × fat signal / (water signal + fat signal). Division was performed where the combined signal was positive. Whole-region summaries included only those valid pixels. For the TA analysis, values outside the mask were replaced with NaN. The class implementation initializes nonpositive-denominator pixels to zero; an additional check confirmed that no TA pixels had nonpositive denominators in any supplied slice.

```python
denominator = water + fat
valid_pixels = denominator > 0

fat_fraction = np.divide(
    fat,
    denominator,
    out=np.zeros_like(fat, dtype=float),
    where=valid_pixels
) * 100

# Keep the whole supplied image region for the first analysis.
fat_fraction_full = fat_fraction.copy()

# Then apply the TA mask for the muscle-specific analysis.
fat_fraction_masked = fat_fraction.copy()
fat_fraction_masked[~mask] = np.nan

whole_image_values = fat_fraction_full[valid_pixels]
muscle_values = fat_fraction_masked[~np.isnan(fat_fraction_masked)]

print(f"Whole supplied image region — mean: {np.mean(whole_image_values):.1f}%")
print(f"Whole supplied image region — minimum: {np.min(whole_image_values):.1f}%")
print(f"Whole supplied image region — maximum: {np.max(whole_image_values):.1f}%")
print()
print(f"TA muscle only — mean: {np.mean(muscle_values):.1f}%")
print(f"TA muscle only — minimum: {np.min(muscle_values):.1f}%")
print(f"TA muscle only — maximum: {np.max(muscle_values):.1f}%")
```

# 4. Visualizing the muscle and fat-fraction map

Water and fat images were displayed with the TA boundary. Whole-region and cropped TA fat-fraction maps used the same 0–100% scale. Cropping changed the display only and did not change the region used for statistics. The cropped panel uses pixel coordinates relative to the crop.

```python
# Find the bounding box of the TA mask
y_pixels, x_pixels = np.where(mask)

y0, y1 = y_pixels.min(), y_pixels.max() + 1
x0, x1 = x_pixels.min(), x_pixels.max() + 1

# Crop the TA-specific fat-fraction map
fat_fraction_TA = fat_fraction_masked[y0:y1, x0:x1]

fig, axes = plt.subplots(1, 4, figsize=(15, 4))

# Water image with TA mask outline
axes[0].imshow(water, cmap="gray")
axes[0].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[0].set_title("Water image + TA mask")

# Fat image with TA mask outline
axes[1].imshow(fat, cmap="gray")
axes[1].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[1].set_title("Fat image + TA mask")

# Fat fraction for the whole supplied image region
whole_image = axes[2].imshow(
    fat_fraction_full,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[2].contour(mask, levels=[0.5], colors="cyan", linewidths=2)
axes[2].set_title("Fat fraction: whole region")

# Zoomed-in fat fraction for the TA only
ta_image = axes[3].imshow(
    fat_fraction_TA,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[3].set_title("TA fat fraction")

for ax in axes:
    ax.set_xlabel("Horizontal pixel position")
    ax.set_ylabel("Vertical pixel position")

# Both fat-fraction plots use the same 0–100% scale
fig.colorbar(
    ta_image,
    ax=axes[2:],
    fraction=0.046,
    pad=0.04,
    label="Fat fraction (%)"
)

plt.show()
```

# 5. Summarizing the distribution

Overlaid histograms with 30 bins for each region summarized the distribution of whole-region and TA-only pixel values. Raw pixel counts were used, so histogram heights depend on the number of pixels in each region. The spread of values provides information not captured by the mean alone.

```python
plt.figure(figsize=(7, 4))
plt.hist(whole_image_values, bins=30, alpha=0.55, label="Whole supplied region")
plt.hist(muscle_values, bins=30, alpha=0.65, label="TA muscle only")
plt.xlabel("Fat fraction (%)")
plt.ylabel("Number of pixels")
plt.title("Fat-fraction distributions")
plt.legend()
plt.show()
```

# 6. Summarizing sampled locations

The same function was applied to all eight TP1 slices. At each location, the mean, median, and population standard deviation (NumPy default ddof=0) of TA fat fraction were calculated, excluding NaN values. Muscle-pixel count was the sum of the Boolean mask. Standard deviation is expressed in percentage points. Each row retained subject, visit, slice ID, percentage location, and physical distance. These are cross-sectional summaries; no volume-weighted whole-muscle average was calculated.

```python
def calculate_fat_fraction(water, fat, mask):
    """Return a masked fat-fraction image in percent."""
    denominator = water + fat
    result = np.divide(
        fat,
        denominator,
        out=np.zeros_like(fat, dtype=float),
        where=denominator > 0
    ) * 100
    result[~mask] = np.nan
    return result

analysis_timepoint = "TP1"  # analyze one visit at a time
analysis_manifest = manifest[manifest["timepoint"] == analysis_timepoint]
results = []

for _, row in analysis_manifest.iterrows():
    water = np.load(DATA_DIR / row["water_file"])
    fat = np.load(DATA_DIR / row["fat_file"])
    mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
    ff = calculate_fat_fraction(water, fat, mask)

    results.append({
        "subject": row["subject"],
        "timepoint": row["timepoint"],
        "slice_id": row["slice_id"],
        "percent_along_muscle": row["percent_along_muscle"],
        "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
        "mean_fat_fraction_percent": np.nanmean(ff),
        "median_fat_fraction_percent": np.nanmedian(ff),
        "std_fat_fraction_percent": np.nanstd(ff),
        "n_muscle_pixels": int(mask.sum()),
    })

results_df = pd.DataFrame(results)
results_df
```

# 7. Optional comparison across visits

The class function was repeated for TP1 and TP2. Mean TA fat fraction was plotted against distance_from_muscle_start_mm because physical sampling intervals were not uniform and differed between visits. Connecting lines visualize the sampled profiles; they are not additional measurements or a separately calculated interpolation dataset.

```python
def summarize_visit(timepoint):
    visit_results = []
    visit_manifest = manifest[manifest["timepoint"] == timepoint]

    for _, row in visit_manifest.iterrows():
        water = np.load(DATA_DIR / row["water_file"])
        fat = np.load(DATA_DIR / row["fat_file"])
        mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
        ff = calculate_fat_fraction(water, fat, mask)

        visit_results.append({
            "timepoint": timepoint,
            "slice_id": row["slice_id"],
            "percent_along_muscle": row["percent_along_muscle"],
            "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
            "mean_fat_fraction_percent": np.nanmean(ff),
        })

    return pd.DataFrame(visit_results)

profiles_df = pd.concat(
    [summarize_visit("TP1"), summarize_visit("TP2")],
    ignore_index=True
)

plt.figure(figsize=(7, 4))
for timepoint, group in profiles_df.groupby("timepoint"):
    group = group.sort_values("distance_from_muscle_start_mm")
    plt.plot(
        group["distance_from_muscle_start_mm"],
        group["mean_fat_fraction_percent"],
        marker="o",
        label=timepoint
    )
plt.xlabel("Distance from muscle start (mm)")
plt.ylabel("Mean TA fat fraction (%)")
plt.title("Mean fat fraction along the tibialis anterior")
plt.legend(title="Visit")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
comparison_source = profiles_df if "profiles_df" in globals() else results_df
comparison = comparison_source.pivot(
    index="percent_along_muscle",
    columns="timepoint",
    values="mean_fat_fraction_percent"
)

if {"TP1", "TP2"}.issubset(comparison.columns):
    comparison["change_TP2_minus_TP1"] = comparison["TP2"] - comparison["TP1"]
    print(comparison.to_string())
else:
    print("Run the optional two-visit extension before making a longitudinal comparison.")
```

# 1. Perform Practice Example

In this section, we copied and pasted from the given example. The following code grabs the values from the Module2_FSHD_Dataset.csv. It then checks for missing values of all subjects and defines T1 and T2 as the earliest year 1 and year 2 visit times. At T1, the MRI fat fraction is interpolated linearly, and at T2, the MRI fat fraction is interpolated quadratically. Then, it graphs the MRI fat fraction from one example subject (13-0004). 

These interpolated values are estimates because they are calculated using existing measurements; it assumes that the MRI fat fraction changes linearly across the first year and quadratically across the second year to define two interpolated points that are in between the measurements, along their corresponding predicted curve. Thus, this method is not extrapolation, which would be estimating beyond the range of sampled data. However, this approach is still limited in that the linear/quadratic curve may not reflect the actual progression of FSHD. For instance, the linear curve assumes there is no upward ceiling to MRI fat fraction that the rate of change decreases towards. Also, FSHD progression may not correlate perfectly with fat fraction of this particular muscle, so it may not be entirely accurate to use these values to represent FSHD progression.


```python
from scipy.interpolate import interp1d #package not imported earlier but necessary for interpolation shortcut

df = pd.read_csv("Module2_FSHD_Dataset.csv")
print(df.shape)
df.head()

# Check for missing visits -- not every subject has a Year 1 and/or Year 2 visit
print("Subjects:", len(df))
print("Have a Year 1 visit:", df['lag_months_y1'].notna().sum())
print("Have a Year 2 visit:", df['lag_months_y2'].notna().sum())

T1 = df['lag_months_y1'].min()   # earliest Year-1 visit across all subjects
T2 = df['lag_months_y2'].min()   # earliest Year-2 visit across all subjects
print(f"T1 = {T1:.2f} months")
print(f"T2 = {T2:.2f} months")

row = df.loc[df['subject_id'] == '13-0004'].iloc[0]

lag_months = np.array([row['lag_months_baseline'], row['lag_months_y1'], row['lag_months_y2']])
ff_values = np.array([row['ff_baseline_R'], row['ff_y1_R'], row['ff_y2_R']])

print("lag_months:", lag_months)
print("ff_values: ", ff_values)


TEAL = "#1f7a8c"
PINK = "#e0569b"
ORANGE = "#e8821e"


f_linear = interp1d(lag_months, ff_values, kind='linear')
f_quadratic = interp1d(lag_months, ff_values, kind='quadratic')

print("f_linear(T1)   =", f_linear(T1))
print("f_quadratic(T2) =", f_quadratic(T2))

x_curve = np.linspace(lag_months[0], lag_months[-1], 200)
quad_curve = f_quadratic(x_curve)
estimate_T1 = float(f_linear(T1))
estimate_T2 = float(f_quadratic(T2))

plt.figure(figsize=(7.5, 5))
plt.plot(x_curve, quad_curve, color=ORANGE, lw=2.5, linestyle="--", label="Quadratic (all 3 points)")
plt.plot(lag_months[:2], ff_values[:2], color=PINK, lw=2.5, linestyle="--", label="Linear (baseline -> Year 1)")
plt.scatter(lag_months, ff_values, color=TEAL, s=90, zorder=4, label="Measured visits (known)")
plt.scatter([T1], [estimate_T1], color=PINK, marker="D", s=120, zorder=5, label=f"Linear estimate @ T1={T1:.1f}")
plt.scatter([T2], [estimate_T2], color=ORANGE, marker="D", s=120, zorder=5, label=f"Quadratic estimate @ T2={T2:.1f}")
plt.title("Subject 13-0004 — right tibialis anterior fat fraction")
plt.xlabel("Months since baseline (lag_months)")
plt.ylabel("Fat fraction (%)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

results = [] 
```

# 2. Interpolating MRI Fat Fraction and Clincial Severity Scores Temporally for All Subjects

This code follows the instructions from the interpolation example file. It interates through a loop to pull the clinical severity scores and fat fraction values from all subjects. In this interpolation method, the rate of change of MRI fat fraction of the tibialis anterior muscle and clinical severity score is assumed to be quadratic. However, if there are only two measurements instead of three, the code interpolates them linearly. The code then subtracts the difference between baseline and timepoints for comparing the consistency of change in the next section #3.

According to the paper, the clinical severity score operates on a 10 point scale: 0 meaning no symptoms and 10 meaning wheelchair bound. According to "Outcome Measures in Facioscapulohumeral Muscular Dystrophy Clinical Trials" by Mehdi Ghasemi and colleagues, the clinical severity score is limited because it "assumes a fixed sequence of muscle involvement and is less suitable for patients with an atypical disease course."

```python

for idx, row in df.iterrows():
    subject_id = row['subject_id']

    # TODO: pull this subject's lag_months and ff_R values into arrays,
    #       the same way we did for Subject 13-0004 above.
    lag_months = np.array([row['lag_months_baseline'], row['lag_months_y1'], row['lag_months_y2']])
    ff_values = np.array([row['ff_baseline_R'], row['ff_y1_R'], row['ff_y2_R']])
    
    css_values = np.array([
    row['clinical_severity_score_baseline'],
    row['clinical_severity_score_y1'],
    row['clinical_severity_score_y2']
    ])


    # TODO: if this subject has at least baseline + Year 1 (no NaNs), compute

    #       their linear estimate at T1 (hint: interp1d with kind='linear').
    #       If they don't have enough data, use NaN instead.
    new_lag_months = []
    new_ff_values = []
    new_css_values = []

    for i in range(len(lag_months)):
        if not np.isnan(lag_months[i]) and not np.isnan(ff_values[i]) and not np.isnan(css_values[i]):
            new_lag_months.append(lag_months[i])
            new_ff_values.append(ff_values[i])
            new_css_values.append(css_values[i])


    lag_months = np.array(new_lag_months)
    ff_values = np.array(new_ff_values)
    css_values = np.array(new_css_values)
    

    # TODO: if this subject has all three visits (no NaNs), compute their
    #       quadratic estimate at T2 (hint: interp1d with kind='quadratic').
    #       If they don't, use NaN instead.

    estimate_T1 = np.nan #initializes the estimate as missing
    estimate_css_T1 = np.nan

    if len(lag_months) >= 2: #if there's at least two points, create the linear interpolation and overwrite the missing variable
        f_linear = interp1d(
            lag_months, ff_values, kind='linear'
        )
        css_linear = interp1d(lag_months, css_values, kind='linear')
        estimate_css_T1 = float(css_linear(T1))
        estimate_T1 = float(f_linear(T1))


    estimate_T2 = np.nan #initializes the estimate as missing
    estimate_css_T2 = np.nan

    if len(lag_months) == 3: ##if there's three points, create the quadratic interpolation and overwrite the missing variable
        f_quadratic = interp1d(
            lag_months, ff_values, kind='quadratic'
        )
        estimate_T2 = float(f_quadratic(T2))
        css_quadratic = interp1d(lag_months, css_values, kind='quadratic')
        estimate_css_T2 = float(css_quadratic(T2))

    change_ff_T1 = estimate_T1 - row['ff_baseline_R'] #solving the research problem of which variable changes most consistently
    change_ff_T2 = estimate_T2 - row['ff_baseline_R']#creates new set of variables that are the differences between measurement and baseline
    change_css_T1 = estimate_css_T1 - row['clinical_severity_score_baseline']
    change_css_T2 = estimate_css_T2 - row['clinical_severity_score_baseline']
    # TODO: append a dict like
    #       {'subject_id': subject_id, 'estimate_T1_linear': ..., 'estimate_T2_quadratic': ...}
    #       to `results`
    results.append({
        'subject_id': subject_id,
        'estimate_T1_linear': estimate_T1,
        'estimate_T2_quadratic': estimate_T2,
        'estimate_css_T1': estimate_css_T1,
        'estimate_css_T2': estimate_css_T2,
        'change_ff_T1': change_ff_T1,
        'change_ff_T2': change_ff_T2,
        'change_css_T1': change_css_T1,
        'change_css_T2': change_css_T2
    })
pass

# TODO: turn `results` into a DataFrame and inspect it
# results_df = pd.DataFrame(results)
# results_df.head()
results_df = pd.DataFrame(results)
print(results_df.head())


# TODO: save your results to a new .csv file, e.g.:
# results_df.to_csv('interpolated_estimates.csv', index=False)
results_df.to_csv('interpolated_estimates.csv', index=False)

```

# 4. Plot the Standard Deviation, Mean Absolute Change, and Percentage Showing Increase Across the Two Biometrics 

Here, we want to answer our project question about which measurement changes most consistently and sensitively. Standard Deviation from baseline measures the varaiblity of the same across the mean to show consistency. As to how sensitive the metrics are to change, we graphed the mean absolute change across the time points from baseline. Interestingly, we also decided to plot the percentage showing an increase from one time interval to the next because that would help determine if the biometric is being sensitive and consistent to FSHD progression.

```python

ff_cols = ['change_ff_T1', 'change_ff_T2']
css_cols = ['change_css_T1', 'change_css_T2']

x = np.arange(2)
width = 0.35

# Calculate statistics
ff_std = results_df[ff_cols].std().values
css_std = results_df[css_cols].std().values

ff_abs = results_df[ff_cols].abs().mean().values
css_abs = results_df[css_cols].abs().mean().values

# Calculate changes between consecutive time points
ff_T1 = results_df['change_ff_T1']
ff_T2 = results_df['change_ff_T2']

css_T1 = results_df['change_css_T1']
css_T2 = results_df['change_css_T2']

# First interval: T1 - baseline
ff_first = ff_T1.dropna()
css_first = css_T1.dropna()

# Second interval: T2 - T1
# Both change columns already subtract the same baseline,
# so subtracting them gives T2 - T1.
ff_second = (ff_T2 - ff_T1).dropna()
css_second = (css_T2 - css_T1).dropna()

ff_pct = [
    (ff_first > 0).mean() * 100,
    (ff_second > 0).mean() * 100
]

css_pct = [
    (css_first > 0).mean() * 100,
    (css_second > 0).mean() * 100
]


def make_graph(ff, css, title, ylabel, ylim=None, xlabels = None): #created a function so wouldn't have to repeat this three more times
    
    if xlabels is None:
        xlabels = ['T1', 'T2']

    plt.figure(figsize=(7, 5))

    bars1 = plt.bar(x - width/2, ff, width,
                    label='MRI Fat Fraction')
    bars2 = plt.bar(x + width/2, css, width,
                    label='Clinical Severity')

    plt.bar_label(bars1, fmt='%.2f', padding=3)
    plt.bar_label(bars2, fmt='%.2f', padding=3)

    plt.xticks(x, xlabels)
    plt.title(title)
    plt.ylabel(ylabel)

    if ylim is not None:
        plt.ylim(0, ylim)
    else:
        plt.ylim(0, max(max(ff), max(css)) * 1.2)

    plt.legend()
    plt.grid(axis='y', alpha=0.2)
    plt.tight_layout()
    plt.show()

# Graph 1
make_graph(ff_std, css_std,
           'Comparing Standard Deviation of Changes from Baseline: Fat Fraction vs. Clinical Severity Score',
           'Standard Deviation of Metric Values from Baseline')

# Graph 2
make_graph(ff_abs, css_abs,
           'Comparing Mean Absolute Change from Baseline: Fat Fraction vs. Clinical Severity Score',
           'Mean Absolute Change of Metric Values from Baseline')

# Graph 3
make_graph(ff_pct, css_pct,
           'Comparing the Percentage of Patients Showing an Increase Across Time Points: \nFat Fraction vs. Clinical Severity Score',
           'Patients with Increase (%)',
           ylim=100,
           xlabels=['Baseline → T1', 'T1 → T2'])
```

# 5. Spatial Interpolation of Fat Fraction Across the Muscle

For our analysis, we decided to only include temporal interpolation of our values, relying on the papers' dataset for their wholistic fat-fraction calculation across the 8 slices. Nevertheless, for practice, here's our code for interpolating the muscle spatially, across 8 slices from the patient UW_0020 analyzed earlier. These slices are at distances 24mm, 44mm, 68mm, 92mm, 116mm, 136mm, 160mm, 184mm for T1. In this example, we performed linear interpolation at 80 mm, and so the values at distance 68 mm and 92mm were used to interpolate the fat fraction of 80 mm as linearly between the two measurements. The variable used to define location along the muscle is our x-axis label of distance from muscle start in mm.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

# Load fat fraction measurements from the 8 MRI slices
spatial_df = pd.read_csv("TA_fat_fraction_summary.csv")

# Get positions and fat fraction measurements
positions = spatial_df['distance_from_muscle_start_mm'].to_numpy()
ff_values = spatial_df['mean_fat_fraction_percent'].to_numpy()

# Create linear interpolation
f_linear = interp1d(positions, ff_values, kind='linear')

# Estimate fat fraction at an unmeasured location
target = 80
estimate = float(f_linear(target))

print(f"Estimated fat fraction at {target} mm: {estimate:.2f}%")

# Plot measured and interpolated values
x_curve = np.linspace(positions.min(), positions.max(), 200)

plt.figure(figsize=(8, 5))
plt.scatter(positions, ff_values, label='Measured MRI slices')
plt.plot(x_curve, f_linear(x_curve), label='Linear interpolation')
plt.scatter(target, estimate, color='red', s=100,
            label=f'Estimated at {target} mm')

plt.xlabel('Distance from muscle start (mm)')
plt.ylabel('Mean fat fraction (%)')
plt.title('Spatial Interpolation Along the Tibialis Anterior')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## Conclusions:

Our analysis and interpolation suggest that MRI fat fraction may detect FSHD progression more consistently than clinical severity score, defined as a clinical rating of overall disease involvement and progression. Mean fat fraction increased by approximately 0.93 percentage points at T1 and 1.52 at T2, while average clinical severity scores changed very little; the mean absolute change mirrors this relationship, indicating that mean fat fraction is a more sensitive measurement to change in FSHD progression. The percent of increase across time points was higher in the MRI fat fraction test than the clinical severity score, which may indicate that the MRI fat fraction is a more consistent metric. Remarkably, the shapes of the two graphs that plotted the mean values of each test looked very similar. However, the graph that plotted the percent increase in MRI analysis also allowed assessment of individual muscle locations, which severity scores cannot directly evaluate. With that being said, larger absolute changes or standard deviations do not establish greater sensitivity because the measures use different scales, and absolute change includes both increases and decreases. These findings support MRI fat fraction as a positive exploratory clinical-trial endpoint. Before using MRI fat fraction to evaluate treatment effectiveness, larger studies would need to establish measurement reliability, validate the interpolated estimates, and show that MRI changes correspond to changes in strength and daily function. Regarding potential ethical implications, relying on MRI alone could overlook benefits patients experience in everyday life like walking or climbing stairs and disadvantage those with limited access to imaging such as geographic areas with less funding or fiscal flexibility.
